### Data Cleaning

```python
# Checking for the number of missing values
df.isna().sum()
```

```python
# Checking for rows with missing 'review_title'
df[df['review_title'].isna()]
```

```python
# Handling duplicates

number_of_duplicates = df.duplicated().sum()
print(f"Number of duplicates: {number_of_duplicates}")

df = df.drop_duplicates()
print("Duplicates removed")
```

```python
# Checking for missing values after removing duplicates
df.isna().sum()
```

```python
# Checking for the row with the missing review_title
df[df['review_title'].isna()]
```

```python
# Concatenating 'review_title' and 'review_text' into 'review_text', handling missing titles.

df['review_text'] = df.apply(
    lambda row: f"{row['review_title']}. {row['review_text']}"
    if pd.notna(row['review_title'])
    else row['review_text'],
    axis=1
)
```

```python
# Convert rating to integer
df['reviewer_rating'] = df['reviewer_rating'].apply(lambda x: int(x[0]))
```

### Final dataset

```python
final_df = df.drop(["reviewer_name", "review_date", "review_title"], axis=1)
```
