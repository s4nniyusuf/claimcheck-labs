### Importing Required Libaries

### Loading the dataset

```python
df = pd.read_csv("/content/drive/MyDrive/data/raw/products.csv")
```

### Data Cleaning

```python
# Checking for the number of missing values per column
df.isna().sum()
```

```python
# Droping rows with no description
df = df.dropna(subset="description")
```

```python
# Handling duplicates
number_of_duplicate_rows = df.duplicated().sum()
print(f"Number of duplicate rows: {number_of_duplicate_rows}")

df = df.drop_duplicates()
print("Duplicates removed")
```

```python
df.info()
```

```python
df['rating'] = df['rating'].apply(lambda x: float(x.split('/')[0]))
```

```python
df.head(1)
```

### Final Dataframe

```python
final_df = df.drop(['url', 'current_price', 'old_price', 'image_url'], axis=1)
```
